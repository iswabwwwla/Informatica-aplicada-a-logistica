```python
import pandas as pd

url = "https://raw.githubusercontent.com/Sainathmudigonda/Student-Result-Analysis/refs/heads/main/student_exam_scores.csv"

df = pd.read_csv(url)

display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Calcular os coeficientes da regressão linear (y = mx + b)
x = df['hours_studied']
y = df['exam_score']
m, b = np.polyfit(x, y, 1)

# Imprimir a equação encontrada
print(f"Equação da Regressão Linear: exam_score = {m:.2f} * hours_studied + {b:.2f}")

# Configurar o estilo do gráfico
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Criar o gráfico de dispersão
sns.scatterplot(data=df, x='hours_studied', y='exam_score', color='blue', alpha=0.7, s=100, label='Dados Reais')

# Adicionar a linha de regressão ao gráfico
x_valores = np.linspace(x.min(), x.max(), 100)
y_valores = m * x_valores + b
plt.plot(x_valores, y_valores, color='red', linewidth=2.5, label=f'Reta de Regressão: y = {m:.2f}x + {b:.2f}')

# Adicionar títulos e rótulos
plt.title('Regressão Linear: Horas Estudadas vs Nota do Exame', fontsize=14, fontweight='bold')
plt.xlabel('Horas Estudadas', fontsize=12)
plt.ylabel('Nota do Exame', fontsize=12)
plt.legend()

# Exibir o gráfico
plt.show()
```
